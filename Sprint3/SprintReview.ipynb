```polyglot-notebook
interface Iinstallable
{
    public void Install();
}

abstract class Component
{
    public int Price { get; private set; }

    public Component(int Price)
    {
        this.Price = Price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Price={Price}");
    }

    public override string ToString()
    {
        return $"Price={Price}";
    }
}

abstract class InstallableComponent : Component, Iinstallable
{
    public InstallableComponent(int Price) : base(Price) {}
    public abstract void Install();
}

class CPU : InstallableComponent
{
    public int CoreCount { get; private set; }

    public CPU(int Price, int CoreCount) : base(Price)
    {
        this.CoreCount = CoreCount;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"{ToString()};CoreCount={CoreCount}");
    }

    public override void Install()
    {
        Console.WriteLine($"Установка CPU");
    }

    public static bool operator ==(CPU FirstCPU, CPU SecondCPU)
    {
        return (FirstCPU.CoreCount == SecondCPU.CoreCount);
    }
    
    public static bool operator !=(CPU FirstCPU, CPU SecondCPU)
    {
        return !(FirstCPU.CoreCount == SecondCPU.CoreCount);
    }
}

class RAM : InstallableComponent
{
    public string DDR_Type { get; set; }

    public RAM(int Price, string DDR_Type) : base(Price)
    {
        this.DDR_Type = DDR_Type;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"{ToString()};DDR_Type={DDR_Type}");
    }

    public override void Install()
    {
        Console.WriteLine("Установка RAM");
    }

    public static bool operator ==(RAM FirstRAM, RAM SecondRAM)
    {
        return (FirstRAM.DDR_Type == SecondRAM.DDR_Type);
    }
    
    public static bool operator !=(RAM FirstRAM, RAM SecondRAM)
    {
        return !(FirstRAM.DDR_Type == SecondRAM.DDR_Type);
    }
}

class GPU : InstallableComponent
{
    public string DirectX_Version { get; set; }
    public int GB_VideoMemory { get; set; }

    public GPU(int Price, string DirectX_Version, int GB_VideoMemory) : base(Price)
    {
        this.DirectX_Version = DirectX_Version;
        this.GB_VideoMemory = GB_VideoMemory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"{ToString()};DirectX_Version={DirectX_Version};GB_VideoMemory={GB_VideoMemory}");
    }

    public override void Install()
    {
        Console.WriteLine("Установка GPU");
    }

    public static bool operator ==(GPU FirstGPU, GPU SecondGPU)
    {
        return (FirstGPU.DirectX_Version == SecondGPU.DirectX_Version &&
        FirstGPU.GB_VideoMemory == SecondGPU.GB_VideoMemory);
    }
    
    public static bool operator !=(GPU FirstGPU, GPU SecondGPU)
    {
        return !(FirstGPU.DirectX_Version == SecondGPU.DirectX_Version &&
        FirstGPU.GB_VideoMemory == SecondGPU.GB_VideoMemory);
    }
}

class HDD : InstallableComponent
{
    public int MB_ReadSpeed { get; private set; }
    public int MB_WriteSpeed { get; private set; }
    public string InterfaceType { get; private set; }
    public int GB_Space { get; private set; }

    public HDD(int Price, int MB_ReadSpeed, int MB_WriteSpeed, string InterfaceType, int GB_Space) : base(Price)
    {
        this.MB_ReadSpeed = MB_ReadSpeed;
        this.MB_WriteSpeed = MB_WriteSpeed;
        this.InterfaceType = InterfaceType;
        this.GB_Space = GB_Space;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"{ToString()};MB_ReadSpeed={MB_ReadSpeed};MB_WriteSpeed={MB_WriteSpeed};InterfaceType={InterfaceType};GB_Space={GB_Space}");
    }

    public override void Install()
    {
        Console.WriteLine("Установка HDD");
    }

    public static bool operator ==(HDD FirstHDD, HDD SecondHDD)
    {
        return (FirstHDD.MB_ReadSpeed == SecondHDD.MB_ReadSpeed &&
        FirstHDD.MB_WriteSpeed == SecondHDD.MB_WriteSpeed &&
        FirstHDD.InterfaceType == SecondHDD.InterfaceType &&
        FirstHDD.GB_Space == SecondHDD.GB_Space);
    }
    
    public static bool operator !=(HDD FirstHDD, HDD SecondHDD)
    {
        return !(FirstHDD.MB_ReadSpeed == SecondHDD.MB_ReadSpeed &&
        FirstHDD.MB_WriteSpeed == SecondHDD.MB_WriteSpeed &&
        FirstHDD.InterfaceType == SecondHDD.InterfaceType &&
        FirstHDD.GB_Space == SecondHDD.GB_Space);
    }
}

class Computer<T> where T : Component, Iinstallable
{
    private HashSet<T> components { get; set; } 
    
    public Computer()
    {
        components = new HashSet<T>();
    }

    public void AddComponent(T component)
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent(T component)
    {
        components.Remove(component);
    }

    public int GetTotalPrice()
    {
        int price = 0;

        foreach (T component in components)
        {
            price += component.Price;
        }

        return price;
    }

    public void DisplayComponents()
    {
        foreach (var component in components)
        {
            component.GetInfo();
        }
    }
}

static class ComputerFactory<T, T2> where T : Computer<T2> where T2 : Component, Iinstallable
{
    private static HashSet<T> computers = new HashSet<T>();

    public static void AddComputer(T computer)
    {
        computers.Add(computer);
    }

    public static void RemoveComputer(T computer)
    {
        computers.Remove(computer);
    }

    public static void DisplayComputers()
    {
        int counter = 0;
        foreach (T computer in computers)
        {
            counter++;
            Console.WriteLine($"Компьютер №{counter}");
            Console.WriteLine($"TotalPrice={computer.GetTotalPrice()}");
            computer.DisplayComponents();
        }
    }
}
Console.WriteLine("=========================Компоненты=======================");

CPU myCPU = new CPU(10000, 8);
myCPU.GetInfo();
CPU myCPU2 = new CPU(10200, 8);
myCPU2.GetInfo();
CPU myCPU3 = new CPU(10000, 12);
myCPU3.GetInfo();

Console.WriteLine(myCPU == myCPU2);
Console.WriteLine(myCPU == myCPU3);

RAM myRAM = new RAM(2300, "DDR4");
myRAM.GetInfo();
RAM myRAM2 = new RAM(2300, "DDR4");
myRAM2.GetInfo();
RAM myRAM3 = new RAM(2500, "DDR5");
myRAM3.GetInfo();

Console.WriteLine(myRAM == myRAM2);
Console.WriteLine(myRAM == myRAM3);

GPU myGPU = new GPU(20000, "12.1", 6);
myGPU.GetInfo();
GPU myGPU2 = new GPU(20000, "12.1", 6);
myGPU2.GetInfo();
GPU myGPU3 = new GPU(20000, "12.1", 4);
myGPU3.GetInfo();

Console.WriteLine(myGPU == myGPU2);
Console.WriteLine(myGPU == myGPU3);

HDD myHDD = new HDD(3000, 460, 360, "USB 3.0", 2048);
myHDD.GetInfo();
HDD myHDD2 = new HDD(3000, 460, 360, "USB 3.0", 2048);
myHDD2.GetInfo();
HDD myHDD3 = new HDD(3000, 460, 400, "USB 3.0", 2048);
myHDD3.GetInfo();

Console.WriteLine(myHDD == myHDD2);
Console.WriteLine(myHDD == myHDD3);
Console.WriteLine("=========================Компьютер========================");

Computer<InstallableComponent> myComputer = new Computer<InstallableComponent>();

myComputer.AddComponent(myCPU);
myComputer.AddComponent(myRAM);
myComputer.AddComponent(myGPU);
myComputer.AddComponent(myHDD);
Console.WriteLine(myComputer.GetTotalPrice());
myComputer.RemoveComponent(myCPU);
Console.WriteLine(myComputer.GetTotalPrice());
myComputer.AddComponent(myCPU2);
Console.WriteLine(myComputer.GetTotalPrice());

Console.WriteLine("======================ComputerFactory=====================");
Console.WriteLine("ComputerFactory");
Console.WriteLine("\n========================myComputer========================\n");
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.AddComputer(myComputer);
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.DisplayComputers();
Console.WriteLine("========================end===============================\n");
Computer<InstallableComponent> myComputer2 = new Computer<InstallableComponent>();

myComputer2.AddComponent(myCPU);
myComputer2.AddComponent(myRAM2);
myComputer2.AddComponent(myGPU2);
myComputer2.AddComponent(myHDD2);
Console.WriteLine("\n=================myComputer,myComputer2====================\n");
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.AddComputer(myComputer2);
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.DisplayComputers();
Console.WriteLine("========================end===============================\n");

Computer<InstallableComponent> myComputer3 = new Computer<InstallableComponent>();

myComputer3.AddComponent(myCPU3);
myComputer3.AddComponent(myRAM3);
myComputer3.AddComponent(myGPU3);
myComputer3.AddComponent(myHDD3);
Console.WriteLine("\n===========myComputer,myComputer2,myComputer3==============\n");
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.AddComputer(myComputer3);
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.DisplayComputers();
Console.WriteLine("========================end===============================\n");

Console.WriteLine("\n=================myComputer,myComputer3====================\n");
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.RemoveComputer(myComputer2);
ComputerFactory<Computer<InstallableComponent>, InstallableComponent>.DisplayComputers();
Console.WriteLine("========================end===============================\n");
```
