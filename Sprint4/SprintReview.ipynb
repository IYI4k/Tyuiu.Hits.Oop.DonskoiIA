```polyglot-notebook
interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; set; }
}

interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; set; }
}

class Enterprise: IEnterprise
{
    private HashSet<ISupplier> suppliers = new HashSet<ISupplier>();
    private HashSet<IEmployee> employees = new HashSet<IEmployee>();

    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    public void Add(ISupplier supplier)
    {
        AddSupplier(supplier);
    }

    public void Add(IEmployee employee)
    {
        AddEmployee(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        List<ISupplier> Isuppliers = new List<ISupplier>();

        foreach (ISupplier supplier in suppliers)
        {
            Isuppliers.Add(supplier);
        }

        return Isuppliers;
    }

    public List<IEmployee> GetEmployees()
    {
        List<IEmployee> Iemployees = new List<IEmployee>();

        foreach (IEmployee employee in employees)
        {
            Iemployees.Add(employee);
        }

        return Iemployees;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal expenses = 0;

        foreach (Supplier supplier in suppliers)
        {
            expenses += supplier.GetCost();
        }

        foreach (Employee employee in employees)
        {
            expenses += employee.GetSalary();
        }

        return expenses;
    }
}

class Supplier: ISupplier
{
    public string Name { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; set; }

    public string GetName()
    {
        return Name;
    }

    public decimal GetCost()
    {
        return Cost;
    }
}

class Employee: IEmployee
{
    public string FirstName { get; set; }
    public string LastName { get; set; }
    public decimal Salary { get; set; }
    public string Position { get; set; }

    public string GetFullName()
    {
        return $"{FirstName} {LastName}";
    }

    public decimal GetSalary()
    {
        return Salary;
    }
}

Enterprise myEnterprise = new Enterprise();

Supplier mySupplier = new Supplier();
mySupplier.Name = "Газодобыча";
mySupplier.Cost = 20000;
mySupplier.SupplierType = "Поставщик газа";

Supplier mySupplier2 = new Supplier();
mySupplier2.Name = "Электроподача";
mySupplier2.Cost = 30000;
mySupplier2.SupplierType = "Поставщик электричества";

Employee myEmployee = new Employee();
myEmployee.FirstName = "Пётр";
myEmployee.LastName = "Теплов";
myEmployee.Salary = 70000;
myEmployee.Position = "Сантехник";

myEnterprise.Add(mySupplier);
myEnterprise.Add(mySupplier2);
myEnterprise.Add(myEmployee);

Console.WriteLine("Список поставщиков:");
int count = 0;
foreach(ISupplier supplier in myEnterprise.GetSuppliers())
{
    count++;
    Console.WriteLine($"{count}. {supplier.GetName()} - {supplier.SupplierType} - Стоимость услуг: {supplier.GetCost()}");
}

Console.WriteLine("Список сотрудников:");
count = 0;
foreach(IEmployee employee in myEnterprise.GetEmployees())
{
    count++;
    Console.WriteLine($"{count}. {employee.GetFullName()} - Должность: {employee.Position} - Зарплата: {employee.GetSalary()}");
}
Console.WriteLine($"Общие расходы: {myEnterprise.CalculateTotalExpenses()}");
```
